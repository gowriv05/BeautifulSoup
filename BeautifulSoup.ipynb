```python
import requests

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/85.0.4183.121 Safari/537.36"
}

url = "https://en.wikipedia.org/wiki/List_of_highest-grossing_films"
html = requests.get(url, headers=headers).text
html
```

```python
from bs4 import BeautifulSoup
```

```python
soup=BeautifulSoup(html,'html.parser')
soup
```

```python
print(soup.prettify)
```

```python
soup.find('table')
```

```python
table=soup.find('table',class_="wikitable sortable plainrowheaders sticky-header col4right col5center col6center")
table
```

```python
table1=soup.find('table',id='mwBME')
table1
```

```python
headers=table.find('tr').text
headers
```

```python
headers1=table1.find('tr').text
headers1
```

```python
header=table.find('tr')
cols=[th.get_text(strip=True) for th in header if th.get_text(strip=True)]
cols
```

```python
trs=table.find_all('tr')[1:]
row_data=[]
for t_row in trs:
  row=[t_data.get_text(strip=True) for t_data in t_row.find_all(['td','th'])]
  row_data.append(row)
row_data
```

```python
import pandas as pd
df=pd.DataFrame(row_data,columns=cols)
df
```
